```python
import os
import os.path as osp
import shutil
import groio
import sys

sys.path.append('src/')
```

---

```python
initial_dataset_name = 'synthetic_theta'
dataset_name = 'InterfaceDataset'

initial_dataset_path = osp.join(initial_dataset_name)
dataset_path = osp.join('data', dataset_name)
```

```python
os.makedirs(dataset_path)
```

```python
interface_counter = {
    'droplet': 0,
    'doughnut': 0,
    'worm': 0,
    'perforation':0,
    'roll': 0,
    'layer': 0
}

for angle_folder in sorted(os.listdir(initial_dataset_path), reverse=True):
    print(angle_folder)
    for interface_folder in os.listdir(osp.join(initial_dataset_path, angle_folder)):
        # if interface_folder == '.DS_Store':
        #     os.remove(osp.join(initial_dataset_path, angle_folder, '.DS_Store'))
        #     continue
        print(interface_folder)
        os.makedirs(osp.join(dataset_path, 'raw_dir', interface_folder), exist_ok=True)

        for file in os.listdir(osp.join(initial_dataset_path, angle_folder, interface_folder)):
            filename = '_'.join([interface_folder, angle_folder, str(interface_counter[interface_folder])]) + '.gro'
            interface_counter[interface_folder] += 1

            shutil.copy(
                osp.join(initial_dataset_path, angle_folder, interface_folder, file),
                osp.join(dataset_path, 'raw_dir', interface_folder, filename)
            )
```
